```python
#  i will use embedding techniq using open source embedding model
#  and store it in faiss,chromadb etc
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS  
from dotenv import load_dotenv
import os
import getpass
load_dotenv()

```

```python
os.environ["LANGCHAIN_API_KEY"] = "[ENCRYPTION_KEY]"
# os.environ["LANGSMITH_API_KEY"] = getpass.getpass("Enter your LangSmith API key: ")
os.environ["LANGSMITH_TRACING"] = "true"
```

```python
# now i will first apply some data ingestion, 
loader=PyPDFLoader("D:\Gen AI_LANGChain\POM07.pdf",mode="page")
docs = loader.load()
print(len(docs))
print(docs[5].page_content)
```

```python
print(type(docs))
print(type(docs[0]))
print(docs[0])
```

```python
#   text  spliter technique
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=30
)

split_docs = text_splitter.split_documents(docs)


```

```python
print(len(split_docs))
print(split_docs[2].page_content)
```

```python
# #  ,then embedinging, then store in db then check similirity search
# use huggingfce embedding
from langchain_huggingface import HuggingFaceEmbeddings
embedding=HuggingFaceEmbeddings(model="BAAI/bge-small-en")

```

```python
vectorstore = FAISS.from_documents(
    split_docs,
    embedding
)
```

```python
query = "what is Fayol 14 Principles of Management?"

results = vectorstore.similarity_search(
    query,
    k=3
)

for doc in results:
    print(doc.page_content)
```
