# String Output Parser Example
This notebook demonstrates how to use LangChain's StrOutputParser with different LLM providers.

```python
# Import required libraries
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os
```

```python
# Load environment variables
load_dotenv()

# Verify API key is loaded
groq_api_key = os.getenv("GROQ_API_KEY")
if not groq_api_key:
    raise ValueError("GROQ_API_KEY not found in environment variables")
print("API Key loaded successfully")
```

```python
# Initialize the LLM with proper configuration
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # You can also use "mixtral-8x7b-32768" or other models
    temperature=0.7,
    max_tokens=1024,
    api_key=groq_api_key
)

print("LLM initialized successfully")
```

```python
# Create a prompt template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant that provides concise and informative answers."),
    ("user", "Tell me about {topic} in 2-3 sentences.")
])

print("Prompt template created")
```

```python
# Create output parser
output_parser = StrOutputParser()

print("Output parser created") 
```

```python
# Create the chain
chain = prompt | llm | output_parser

print("Chain created successfully")
```

```python
# Invoke the chain
result = chain.invoke({"topic": "langchain"})
print("\nResult:")
print(result)
```

```python
# Try with different topics
topics = ["artificial intelligence", "machine learning", "natural language processing"]

for topic in topics:
    result = chain.invoke({"topic": topic})
    print(f"\n{'='*50}")
    print(f"Topic: {topic}")
    print(f"{'='*50}")
    print(result)
```
