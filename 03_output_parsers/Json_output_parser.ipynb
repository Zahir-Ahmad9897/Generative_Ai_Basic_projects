
Structured Output from LLM using JsonOutputParser

This notebook demonstrates how to use the `JsonOutputParser` in LangChain to get structured JSON data from a Chat Model without using a Pydantic schema.

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate
```

```python
# Load environment variables
load_dotenv(dotenv_path="D:\\Gen AI_LANGChain\\.env")

groq_api_key = os.getenv("GROQ_API_KEY")
if not groq_api_key:
    raise ValueError("GROQ_API_KEY not found in environment variables")

print("Environment variables loaded successfully")
```

### 1. Initialize Model and Parser


```python
# Using a supported model and fixing 'temperature' typo
model = ChatGroq(model_name="llama-3.3-70b-versatile", temperature=0.2)

# Initializing without Pydantic as requested
parser = JsonOutputParser()
```

### 2. Create the Prompt
When not using Pydantic, the `format_instructions` will be more generic, so we should be clear in our template.

```python
prompt = ChatPromptTemplate.from_template(
    template="""give me result with the name, age, and city of {topic}.
{format_instructions}"""
)

prompt = prompt.partial(format_instructions=parser.get_format_instructions())
```

### 3. Run the Chain

```python
chain = prompt | model | parser

result = chain.invoke({"topic": "a fictional character named Alex"})
print(result)
```
