```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
loader = WebBaseLoader(web_path="https://python.langchain.com/docs/integrations/document_loaders/#webpages",
                       bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                           class_=("docsWrapper_hBAB")
                       ))
                       )
doc=loader.load()
# print(doc[0])
```

```python
from requests_html import AsyncHTMLSession

asession = AsyncHTMLSession()

async def get_content():
    r = await asession.get("https://python.langchain.com/docs/integrations/document_loaders/#webpages")
    await r.html.arender(timeout=20)
    content = r.html.find("main", first=True).text
    print(content[:1000])  # Print first 1000 characters

# Run async function safely in notebook
import asyncio
await get_content()
whrer from webscrating as i say so  
```
